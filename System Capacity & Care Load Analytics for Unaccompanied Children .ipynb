```python
hp["Children discharged from HHS Care"] = hp["Children discharged from HHS Care"].astype(float)
```

```python
hp["Children in HHS Care"]=hp["Children in HHS Care"].str.replace(",", "", regex=False).astype(float)
```

```python
hp["system load"]= hp["Children in CBP custody"] + hp["Children in HHS Care"]
```

```python
hp["Net daily intake"]= hp["Children transferred out of CBP custody"] -  hp["Children discharged from HHS Care"]
```

```python
hp["care load"] = hp["system load"].pct_change()*100
```

```python
hp["system load"].dtype
```

```python
hp["Backlog indicator"] = hp["Net daily intake"]>0
```

```python
hp.head(10)
```

```python
hp.tail(10)
```

```python
hp["high load"] = hp["system load"] > hp["system load"].mean()
```

```python
hp["high load"].value_counts()
```

```python
hp["system load"].mean()
```

```python
hp["High_Load_Group"] = (
    hp["high load"] != hp["high load"].shift()
).cumsum()
```

```python
hp["High_Load_Duration"] = (
    hp.groupby("High_Load_Group")["high load"]
    .transform("sum"))
```

```python
hp["Sustained_High_Load"] = (
    hp["high load"] &
    (hp["High_Load_Duration"] >= 3)
)
```

```python
hp["Sustained_High_Load"].value_counts()
```

```python
hp["7_day_avg"] = hp["system load"].rolling(7,min_periods = 1).mean()
```

```python
hp["14_day_avg"]= hp["system load"].rolling(14,min_periods= 1).mean()
```

```python
hp.head(30)
```

```python
hp["system load"].notna().sum()
```

```python
hp["Strain_Group"] = (
    hp["Sustained_High_Load"] != hp["Sustained_High_Load"].shift()
).cumsum()
```

```python
hp["Strain_Duration"] = (
    hp.groupby("Strain_Group")["Sustained_High_Load"]
      .transform("sum")
)
```

```python
hp["Prolonged_Strain"] = (
    hp["Sustained_High_Load"] &
    (hp["Strain_Duration"] >= 3)
)
```
